```python
import math

while True:

    # 첫 번째 숫자 입력
    while True:
        try:
            num1 = int(input("첫 번째 정수를 입력하세요: "))
            break
        except ValueError:
            print("정수를 입력해주세요.")

    # 두 번째 숫자 입력
    while True:
        try:
            num2 = int(input("두 번째 정수를 입력하세요: "))
            break
        except ValueError:
            print("정수를 입력해주세요.")

    op = input("연산자(+,-,*,/,**)를 입력하세요: ")

    try:
        if op == "+":
            result = num1 + num2

        elif op == "-":
            result = num1 - num2

        elif op == "*":
            result = num1 * num2

        elif op == "/":
            result = num1 / num2

        elif op == "**":
            result = math.pow(num1, num2)

        else:
            print("지원하지 않는 연산자입니다.")
            continue

        print("결과:", result)

    except ZeroDivisionError:
        print("0으로 나눌 수 없습니다.")
        continue

    again = input("계속 계산하시겠습니까? (y/n): ")

    if again.lower() != "y":
        print("계산기를 종료합니다.")
        break
```
